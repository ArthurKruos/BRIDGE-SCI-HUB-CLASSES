### $\text{Pip e Ambientes Virtuais (venv)}$

Quando trabalhamos localmente (fora do Colab), precisamos gerenciar nossas ferramentas:

1.  **Pip ($pip$):** É o instalador de pacotes do Python.  
    $\text{Exemplo: } pip \ install \ pandas$
2.  **Ambiente Virtual ($venv$):** É uma pasta isolada para cada projeto. Imagine que o Projeto A usa uma versão antiga do Pandas e o Projeto B usa uma nova; o $venv$ evita que um quebre o outro.

**Como criar no seu computador (Terminal):**
1. Criar o ambiente: `python -m venv venv`
2. Ativar (Windows): `.\venv\Scripts\activate`
3. Ativar (Mac/Linux): `source venv/bin/activate`

## Variáveis e Tipos de Dados

```python
# No mercado financeiro, precisamos guardar informações.
# Variáveis são como "etiquetas" para esses dados.

ativo = "PETR4"          # String (Texto)
preco_atual = 34.50      # Float (Número decimal)
quantidade_lote = 100    # Int (Número inteiro)
esta_em_alta = True      # Boolean (Verdadeiro ou Falso)

print(f"O ativo {ativo} está custando R$ {preco_atual}")

# Listas: Guardando vários itens juntos (Ex: Uma carteira ou lista de países)
paises_brics = ["Brasil", "Rússia", "Índia", "China", "África do Sul"]
precos_das_acoes = [25.40, 32.10, 45.00, 19.80]

print("Países no radar:", paises_brics)
```

## Condicionais e Repetição

```python
# Estrutura IF/ELSE: Tomada de decisão
preco_alvo = 40.00

if preco_atual < preco_alvo:
    print("O ativo está abaixo do preço alvo. Oportunidade de compra?")
else:
    print("O ativo atingiu ou superou o preço alvo.")

print("-" * 30)

# Estrutura FOR: Automação de tarefas repetitivas
print("Lista de preços da carteira:")
for preco in precos_das_acoes:
    # Calculando um imposto hipotético de 15% sobre cada preço
    preco_com_imposto = preco * 1.15
    print(f"Preço original: {preco} | Com imposto: {preco_com_imposto:.2f}")
```

## Introdução ao Pandas

```python
import pandas as pd

# O DataFrame é a "Planilha" do Python.
# Vamos criar uma tabela simples para entender a estrutura
dados_bridge = {
    'Ticker': ['VALE3', 'ITUB4', 'ABEV3', 'BBDC4'],
    'Setor': ['Mineração', 'Bancos', 'Consumo', 'Bancos'],
    'Preço': [68.50, 32.10, 14.20, 13.90],
    'Variação_Dia_%': [1.2, -0.5, 0.8, -1.1]
}

df = pd.DataFrame(dados_bridge)

# Visualização inicial
print("Visualizando o DataFrame:")
display(df.head()) # Mostra as primeiras linhas de forma organizada
```

## Exploração e Filtros

```python
# Como extrair informações rápidas da base de dados
print("Resumo estatístico dos preços:")
display(df['Preço'].describe())

print("\nFiltrando apenas ativos do setor de Bancos:")
bancos = df[df['Setor'] == 'Bancos']
display(bancos)

print("\nAtivos com variação positiva:")
positivos = df[df['Variação_Dia_%'] > 0]
display(positivos)
```
